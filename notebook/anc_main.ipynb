```python
def check_date_validity(df, col, earliest, latest):
    earliest_date = pd.to_datetime(earliest)
    latest_date = pd.to_datetime(latest)
    date_series = pd.to_datetime(df[col], errors="coerce")
    non_null_mask = date_series.notna()
    total_non_null = non_null_mask.sum()
    if total_non_null == 0:
        return 0, 0
    valid = ((date_series[non_null_mask] >= earliest_date) & (date_series[non_null_mask] <= latest_date)).sum()
    return total_non_null, int(valid)

def check_validity_numeric(df, col, min_val=12, max_val=55):
    number = pd.to_numeric(df[col], errors="coerce")
    non_null_mask = number.notna()
    total_non_null = non_null_mask.sum()
    if total_non_null == 0:
        return 0, 0
    valid = number[non_null_mask].between(min_val, max_val, inclusive="both").sum()
    return total_non_null, int(valid)

def check_validity_categorical(df, col, valid_values):
    valid_set = set(valid_values)
    series = df[col].replace("", pd.NA)
    non_null_mask = series.notna()
    total_non_null = non_null_mask.sum()
    if total_non_null == 0:
        return 0, 0
    valid = series[non_null_mask].astype(str).isin(valid_set).sum()
    return total_non_null, int(valid)

def check_validity_lmp(df, lmp_col="lmp", date_col="date"):
    visit_date = pd.to_datetime(df[date_col], errors="coerce")
    lmp = pd.to_datetime(df[lmp_col], errors="coerce")
    non_null_mask = lmp.notna() & visit_date.notna()
    total_non_null = non_null_mask.sum()
    if total_non_null == 0:
        return 0, 0
    lmp_nn, visit_nn = lmp[non_null_mask], visit_date[non_null_mask]
    valid = ((lmp_nn <= visit_nn) & (lmp_nn >= (visit_nn - pd.Timedelta(weeks=45)))).sum()
    return total_non_null, int(valid)

def check_validity_edd(df, edd_col="edd", date_col="date"):
    visit_date = pd.to_datetime(df[date_col], errors="coerce")
    edd = pd.to_datetime(df[edd_col], errors="coerce")
    non_null_mask = edd.notna() & visit_date.notna()
    total_non_null = non_null_mask.sum()
    if total_non_null == 0:
        return 0, 0
    edd_nn, visit_nn = edd[non_null_mask], visit_date[non_null_mask]
    valid = ((edd_nn >= visit_nn) & (edd_nn <= (visit_nn + pd.Timedelta(weeks=45)))).sum()
    return total_non_null, int(valid)

def check_validity_pregnancy_weeks(df, col="pregnancy_weeks", min_val=0, max_val=45):
    weeks = pd.to_numeric(df[col].astype(str).str.split().str[0], errors="coerce")
    non_null_mask = weeks.notna()
    total_non_null = non_null_mask.sum()
    if total_non_null == 0:
        return 0, 0
    valid = weeks[non_null_mask].between(min_val, max_val, inclusive="both").sum()
    return total_non_null, int(valid)

def check_validity_yes_no_binary_columns(df, col):
    series = df[col].replace("", pd.NA)
    non_null_mask = series.notna()
    total_non_null = non_null_mask.sum()
    if total_non_null == 0:
        return 0, 0
    valid = series[non_null_mask].astype(str).isin(["Yes", "No", "1", "0"]).sum()
    return total_non_null, int(valid)
```

## 8. Uniqueness for ANC Registration
Uniqueness for ANC Registration was measured on three metric
- To see duplicates in 'name', we expect this to be null as this is system created
- Duplicates in 'patient'
- DUplicates in 'patient' with gravida with the assumption that a patient came back for another pregnancy

```python
def uniqueness_checks_reg(group):
    n = len(group)
    checks = []

    # 1. Duplicate record ID
    dup_name = group.duplicated(subset=['name'], keep=False).sum()
    checks.append(("Duplicate record ID (name)", n, int(dup_name)))

    # 2. Duplicate patient
    sub = group.dropna(subset=['patient'])
    dup_patient = sub.duplicated(subset=['patient'], keep=False).sum()
    checks.append(("Duplicate patient", len(sub), int(dup_patient)))

    # 3. Duplicate patient with same gravida_g
    sub2 = group.dropna(subset=['patient', 'gravida_g'])
    dup_gravida = sub2.duplicated(subset=['patient', 'gravida_g'], keep=False).sum()
    checks.append(("Duplicate patient + gravida_g", len(sub2), int(dup_gravida)))

    return pd.DataFrame(checks, columns=['check', 'total_records', 'duplicates'])

_results = []
for (company, month), group in anc_reg.groupby(['company', 'month_reg'], observed=True):
    result = uniqueness_checks_reg(group)
    result['company'] = company
    result['month_reg'] = month
    _results.append(result)

anc_reg_uniqueness_wc = pd.concat(_results, ignore_index=True)
anc_reg_uniqueness_wc['uniqueness_pct'] = (
    (anc_reg_uniqueness_wc['total_records'] - anc_reg_uniqueness_wc['duplicates'])
    / anc_reg_uniqueness_wc['total_records'].replace(0, np.nan) * 100
).round(2)

anc_reg_uniqueness_wc = anc_reg_uniqueness_wc[['month_reg', 'company', 'check', 'total_records', 'duplicates', 'uniqueness_pct']]
print(f"Shape: {anc_reg_uniqueness_wc.shape}")
```

- alive_children <= parity_with_alive_births
- If previous_cesarean_section = Yes → specify_cesarean_date should be filled


```python
def consistency_checks_reg(group):
    checks = []

    # 1. gravida_g >= parity_with_alive_births + premature_deliveries + number_of_abortions
    sub1 = group.dropna(subset=['gravida_g', 'parity_with_alive_births', 'premature_deliveries', 'number_of_abortions'])
    sub1 = sub1.copy()
    for col in ['gravida_g', 'parity_with_alive_births', 'premature_deliveries', 'number_of_abortions']:
        sub1[col] = pd.to_numeric(sub1[col], errors='coerce')
    sub1 = sub1.dropna(subset=['gravida_g', 'parity_with_alive_births', 'premature_deliveries', 'number_of_abortions'])
    consistent1 = (sub1['gravida_g'] >= sub1['parity_with_alive_births'] + sub1['premature_deliveries'] + sub1['number_of_abortions']).sum()
    checks.append(("gravida_g >= parity + premature + abortions", len(sub1), int(consistent1)))

    # 2. alive_children <= parity_with_alive_births
    sub2 = group.dropna(subset=['alive_children', 'parity_with_alive_births']).copy()
    for col in ['alive_children', 'parity_with_alive_births']:
        sub2[col] = pd.to_numeric(sub2[col], errors='coerce')
    sub2 = sub2.dropna(subset=['alive_children', 'parity_with_alive_births'])
    consistent2 = (sub2['alive_children'] <= sub2['parity_with_alive_births']).sum()
    checks.append(("alive_children <= parity_with_alive_births", len(sub2), int(consistent2)))

    # 3. If previous_cesarean_section = Yes → specify_cesarean_date should be filled
    sub3 = group[group['previous_cesarean_section'] == 'Yes'].copy()
    sub3['specify_cesarean_date'] = sub3['specify_cesarean_date'].replace("", pd.NA)
    consistent3 = sub3['specify_cesarean_date'].notna().sum()
    checks.append(("cesarean=Yes → date filled", len(sub3), int(consistent3)))

    return pd.DataFrame(checks, columns=['check', 'total', 'consistent'])


_results = []
for (company, month), group in anc_reg.groupby(['company', 'month_reg'], observed=True):
    result = consistency_checks_reg(group)
    result['company'] = company
    result['month_reg'] = month
    _results.append(result)

anc_reg_consistency_wc = pd.concat(_results, ignore_index=True)
anc_reg_consistency_wc['consistency_pct'] = (
    anc_reg_consistency_wc['consistent'] / anc_reg_consistency_wc['total'].replace(0, np.nan) * 100
).round(2)

anc_reg_consistency_wc = anc_reg_consistency_wc[['month_reg', 'company', 'check', 'total', 'consistent', 'consistency_pct']]
print(f"Shape: {anc_reg_consistency_wc.shape}")
```

## 12. Consistency ANC FOllowup
For Consistency in ANC Followup, we use the following
- Fundal height ≈ GA weeks (±3 cm) — valid only between 20–36 weeks
- followup_date should be before rendezvous

```python
anc_foll_1['calculated_ga_weeks'] = (
    anc_foll_1['pregnancy_weeks_fol']
    .astype(str)
    .str.extract(r'^(\d+)\s+weeks', expand=False)
    .astype(float)
)
```

```python
def consistency_checks_foll(group):
    checks = []

    # 1. Fundal height ≈ GA weeks (±3 cm) — valid only between 20–36 weeks
    sub1 = group.copy()
    sub1['calculated_ga_weeks'] = pd.to_numeric(sub1['calculated_ga_weeks'], errors='coerce')
    sub1['fundal_height'] = pd.to_numeric(sub1['fundal_height'], errors='coerce')
    sub1 = sub1.dropna(subset=['calculated_ga_weeks', 'fundal_height'])
    sub1 = sub1[sub1['calculated_ga_weeks'].between(20, 36)]
    consistent1 = (abs(sub1['fundal_height'] - sub1['calculated_ga_weeks']) <= 3).sum()
    checks.append(("Fundal height ≈ GA weeks (±3cm, 20–36 wks)", len(sub1), int(consistent1)))

    # 2. followup_date should be before rendezvous
    sub2 = group.copy()
    sub2['followup_date'] = pd.to_datetime(sub2['followup_date'], errors='coerce')
    sub2['rendezvous'] = pd.to_datetime(sub2['rendezvous'], errors='coerce')
    sub2 = sub2.dropna(subset=['followup_date', 'rendezvous'])
    consistent2 = (sub2['followup_date'] < sub2['rendezvous']).sum()
    checks.append(("followup_date < rendezvous", len(sub2), int(consistent2)))

    return pd.DataFrame(checks, columns=['check', 'total', 'consistent'])


_results = []
for (company, month), group in anc_foll_1.groupby(['company', 'month_foll'], observed=True):
    result = consistency_checks_foll(group)
    result['company'] = company
    result['month_foll'] = month
    _results.append(result)

anc_foll_consistency = pd.concat(_results, ignore_index=True)
anc_foll_consistency['consistency_pct'] = (
    anc_foll_consistency['consistent'] / anc_foll_consistency['total'].replace(0, np.nan) * 100
).round(2)

anc_foll_consistency = anc_foll_consistency[['month_foll', 'company', 'check', 'total', 'consistent', 'consistency_pct']]
print(f"Shape: {anc_foll_consistency.shape}")
```

```python
"ANC New Registrations" is count by company and month_reg for df called anc_reg_first
"ANC new registrations under 14 years" where 'age' is below 14
"ANC new registrations aged 18-19 years" where 'age' is between 18-19
"ANC new registrations aged 14-17 years" where 'age' is below 14-17
"ANC First standard contact 1st trimester" where 'calculated_ga_weeks' less than 12
"ANC new registrations received ultrasound scan" where "ultrasound" == "Yes"
"ANC new registrations screened for malnutrition_MUAC" where "muac" is not null
"ANC new registrations screened who were malnourished_MUAC < 21 cm" where "muac" is less than 21
"ANC risk pregnancy detected_including pregnancy under 15 years" where "pregnancy_risk_status" is "Risky"
"ANC new registrations who received full course of Iron and Folic Acid supplements _90 tablets" where "provision_of_iron_folic_acid" is "Yes"
"ANC Insecticide Treated Bed nets distributed" where "provision_of_mosquito_net" is "Yes"
"ANC new registrations who received MMS Supplement" where "provision_of_mms" is "Yes"
"ANC new registrations tested for syphilis" where "rpr" is not na
"ANC new registrations syphilis tested positive" where "rpr" is "Positive"
"ANC new registrations tested for anaemia" where "hemoglobin" is not na and numeric
"ANC new registrations with anemia Hb < 11 g/dL" where "hemoglobin" is not na and numeric and less than 11
"ANC Number of pregnant women screened for HBV" where "hepatite_b" is not na
"ANC Number of pregnant women with HBV infection defined by HBsAg-positive serological status" where "hepatite_b" is "Positive"
```

```python
_df = anc_reg_first.copy()

# Numeric conversions
_df['age'] = pd.to_numeric(_df['age'], errors='coerce')
_df['muac'] = pd.to_numeric(_df['muac'], errors='coerce')
_df['hemoglobin'] = pd.to_numeric(_df['hemoglobin'], errors='coerce')

# calculated_ga_weeks from pregnancy_weeks_fol
_df['calculated_ga_weeks'] = (
    _df['pregnancy_weeks_fol'].astype(str)
    .str.extract(r'^(\d+)\s+weeks', expand=False)
    .astype(float)
)

# Clean string columns used in notna checks
for col in ['rpr', 'hepatite_b']:
    _df[col] = _df[col].replace("", pd.NA)

# Indicator flags
_df['ANC New Registrations'] = 1
_df['ANC new registrations under 14 years']                                                                    = (_df['age'] < 14).astype(int)
_df['ANC new registrations aged 14-17 years']                                                                  = (_df['age'].between(14, 17)).astype(int)
_df['ANC new registrations aged 18-19 years']                                                                  = (_df['age'].between(18, 19)).astype(int)
_df['ANC First standard contact 1st trimester']                                                                = (_df['calculated_ga_weeks'] < 12).astype(int)
_df['ANC new registrations received ultrasound scan']                                                          = (_df['ultrasound'] == 'Yes').astype(int)
_df['ANC new registrations screened for malnutrition_MUAC']                                                    = _df['muac'].notna().astype(int)
_df['ANC new registrations screened who were malnourished_MUAC < 21 cm']                                       = (_df['muac'] < 21).astype(int)
_df['ANC risk pregnancy detected_including pregnancy under 15 years']                                          = (_df['pregnancy_risk_status'] == 'Risky').astype(int)
_df['ANC new registrations who received full course of Iron and Folic Acid supplements _90 tablets']           = (_df['provision_of_iron_folic_acid'] == 'Yes').astype(int)
_df['ANC Insecticide Treated Bed nets distributed']                                                            = (_df['provision_of_mosquito_net'] == 'Yes').astype(int)
_df['ANC new registrations who received MMS Supplement']                                                       = (_df['provision_of_mms'] == 'Yes').astype(int)
_df['ANC new registrations tested for syphilis']                                                               = _df['rpr'].notna().astype(int)
_df['ANC new registrations syphilis tested positive']                                                          = (_df['rpr'] == 'Positive').astype(int)
_df['ANC new registrations tested for anaemia']                                                                = _df['hemoglobin'].notna().astype(int)
_df['ANC new registrations with anemia Hb < 11 g/dL']                                                         = (_df['hemoglobin'] < 11).astype(int)
_df['ANC Number of pregnant women screened for HBV']                                                           = _df['hepatite_b'].notna().astype(int)
_df['ANC Number of pregnant women with HBV infection defined by HBsAg-positive serological status']           = (_df['hepatite_b'] == 'Positive').astype(int)

indicators = [
    'ANC New Registrations',
    'ANC new registrations under 14 years',
    'ANC new registrations aged 14-17 years',
    'ANC new registrations aged 18-19 years',
    'ANC First standard contact 1st trimester',
    'ANC new registrations received ultrasound scan',
    'ANC new registrations screened for malnutrition_MUAC',
    'ANC new registrations screened who were malnourished_MUAC < 21 cm',
    'ANC risk pregnancy detected_including pregnancy under 15 years',
    'ANC new registrations who received full course of Iron and Folic Acid supplements _90 tablets',
    'ANC Insecticide Treated Bed nets distributed',
    'ANC new registrations who received MMS Supplement',
    'ANC new registrations tested for syphilis',
    'ANC new registrations syphilis tested positive',
    'ANC new registrations tested for anaemia',
    'ANC new registrations with anemia Hb < 11 g/dL',
    'ANC Number of pregnant women screened for HBV',
    'ANC Number of pregnant women with HBV infection defined by HBsAg-positive serological status',
]

accuracy_ebuzima = (
    _df.groupby(['company', 'month_reg'], observed=True)[indicators]
    .sum()
    .reset_index()
    .melt(id_vars=['company', 'month_reg'], var_name='indicator', value_name='ebuzima_value')
)[['company', 'month_reg', 'indicator', 'ebuzima_value']]

print(f"Shape: {accuracy_ebuzima.shape}")
```
